# 📊 Análise de Comportamento de Clientes e Gorjetas

Este projeto explora padrões de consumo em um restaurante, analisando fatores que influenciam o valor das gorjetas, como valor da conta, dia da semana e perfil do cliente.

## 🎯 Objetivo

Investigar quais fatores impactam o valor da gorjeta e identificar padrões de comportamento dos clientes que possam auxiliar na tomada de decisão.

## 📁 Dataset

O dataset contém informações sobre:

- Valor da conta (`total_bill`)
- Gorjeta (`tip`)
- Sexo (`sex`)
- Fumante (`smoker`)
- Dia da semana (`day`)
- Período (`time`)
- Número de pessoas (`size`)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='whitegrid')
```

```python
df = pd.read_csv('dataset.csv')
df.head()
```

## 🔎 Visão geral dos dados

A primeira visualização permite entender a estrutura do dataset e identificar as variáveis disponíveis.

```python
df.shape
```

O dataset possui 244 registros e 7 variáveis.

```python
df.info()
```

Não há valores nulos no dataset. Os dados incluem variáveis numéricas e categóricas, permitindo análises de comportamento e comparação entre grupos.

```python
df.describe()
```

A média das contas está próxima de 20, enquanto a gorjeta média gira em torno de 3, indicando um padrão consistente de consumo.

```python
sns.histplot(df['total_bill'], kde=True)
plt.title('Distribuição do valor da conta')
plt.show()
```

A maioria das contas está concentrada entre 10 e 20, com poucos valores mais altos.

```python
sns.boxplot(x='sex', y='tip', data=df)
plt.title('Gorjeta por sexo')
plt.show()
```

Homens apresentam maior variação nas gorjetas, enquanto mulheres mostram comportamento mais consistente.

```python
sns.boxplot(x='day', y='tip', data=df)
plt.title('Gorjeta por dia da semana')
plt.show()
```

Os maiores valores de gorjeta ocorrem nos finais de semana, especialmente aos domingos.

```python
sns.boxplot(x='smoker', y='tip', data=df)
plt.title('Gorjeta: fumantes vs não fumantes')
plt.show()
```

Fumantes apresentam leve tendência a gorjetas maiores, mas sem diferença significativa.

```python
df['tip_pct'] = (df['tip'] / df['total_bill']) * 100
```

```python
sns.histplot(df['tip_pct'], kde=True)
plt.title('Distribuição da porcentagem de gorjeta')
plt.show()
```

A porcentagem de gorjeta permite analisar o comportamento proporcional dos clientes, independente do valor total da conta.

```python
df[['total_bill', 'tip', 'size']].corr()
```

```python
sns.scatterplot(x='total_bill', y='tip', data=df)
plt.title('Relação entre conta e gorjeta')
plt.show()
```

Existe uma relação positiva entre valor da conta e gorjeta. Clientes que gastam mais tendem a dar gorjetas maiores.

## 💡 Conclusão

A análise indica que o valor da gorjeta está mais relacionado ao valor da conta e ao contexto da refeição (como finais de semana) do que a características individuais.

Esses insights podem ser úteis para entender comportamento do consumidor e apoiar decisões em negócios de alimentação e plataformas digitais.
